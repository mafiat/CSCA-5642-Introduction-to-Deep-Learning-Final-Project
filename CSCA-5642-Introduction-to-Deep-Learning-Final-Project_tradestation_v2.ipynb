#### Project Overview: Stock Market Prediction using Deep Learning

This project aims to predict stock market price movements using deep learning techniques. We will focus on predicting the future price of a stock.

Here is a breakdown of the steps involved in this notebook:

**1. Data Collection:**
*   **Source:** We fetch up to 10 years of historical daily stock data using the **Tradestation API**.
*   **Features:** The dataset includes Open, High, Low, Close prices, and Volume (OHLCV).
*   **Target:** We will analyze **SPY** (S&P 500 ETF) and **NVDA** (NVIDIA Corp).

**2. Data Preprocessing and Feature Engineering:**
*   **Data Cleaning:** Handle missing values and rows with zero volume.
*   **Feature Scaling:** Normalize the features to a range of (0, 1) using `MinMaxScaler`, which is essential for stable training of neural networks.
*   **Technical Indicators:** Engineer new features from the OHLCV data to provide more context to the model. The indicators created are:
    *   Simple Moving Averages (SMA)
    *   Relative Strength Index (RSI)
    *   Moving Average Convergence Divergence (MACD)
    *   Bollinger Bands
*   **Sequence Creation:** The time-series data is transformed into sequences. We use a look-back period of 90 days of data to predict the next day's closing price.

**3. Model Selection:**
*   **LSTM (Long Short-Term Memory):** We use an LSTM network, a type of Recurrent Neural Network (RNN) well-suited for time-series forecasting due to its ability to capture long-term dependencies in data.

**4. Model Training:**
*   **Data Splitting:** The data is split chronologically into an 80% training set and a 20% testing set.
*   **Model Architecture:** The LSTM model consists of multiple LSTM layers with `Dropout` for regularization, followed by `Dense` layers to produce a single output value for the price prediction.
*   **Compilation:**
    *   **Loss Function:** We use `Mean Squared Error (MSE)` as the loss function, which is standard for regression tasks like price prediction.
    *   **Optimizer:** The `Adam` optimizer is used to train the model.
*   **Training:** The model is trained for a set number of epochs, with performance monitored on a validation set.

**5. Model Evaluation:**
*   **Price Prediction Metrics:** The model's performance is evaluated on the unseen test set using **Root Mean Squared Error (RMSE)**.
*   **Visualization:** We plot the predicted prices against the actual prices to visually assess how well the model's predictions align with the real market data.

4. Data Fetching

```python
def fetch_data(symbol):
    """
    Fetches historical stock data from TradeStation API.
    """
    print(f"Connecting to TradeStation and fetching data for {symbol}...")
    try:

        lastdate = datetime.now()
        firstdate = lastdate - timedelta(days=365*10)

        # The v3 endpoint for barcharts is now under /marketdata/
        bars_url = f"{API_BASE_URL}/marketdata/barcharts/{symbol}/{TIMEFRAME}?lastdate={lastdate.isoformat()}&firstdate={firstdate.isoformat()}"

        headers = {
            'Authorization': f'Bearer {get_access_token()}'
        }

        try:
            response = requests.request("GET", bars_url, headers=headers)
            response.raise_for_status()

            data = response.json().get('Bars')

            if not data:
                print("No data returned from the API.")
                df = pd.DataFrame()
            else:
                df = pd.DataFrame(data)

                # The v3 API returns 'TimeStamp' and other columns capitalized
                df['timestamp'] = pd.to_datetime(df['TimeStamp'])
                df.set_index('timestamp', inplace=True)
                df.sort_index(inplace=True)
                df = df[['Open', 'High', 'Low', 'Close', 'TotalVolume']]
                df['Close'] = pd.to_numeric(df['Close'], errors='coerce')
                df['TotalVolume'] = pd.to_numeric(df['TotalVolume'], errors='coerce')

        except requests.exceptions.HTTPError as err:
            print(f"HTTP Error: {err}")
            df = pd.DataFrame()
        
        except Exception as e:
            print(f"An error occurred: {e}")
            df = pd.DataFrame()
        
        print("Data fetched successfully.")
        return df
    
    except Exception as e:
        print(f"Error fetching data from TradeStation: {e}")
    return None

    
        
```

5. Data Retrieval and Feature Engineering

```python
def create_features(df):
    """
    Creates several technical indicator features from the raw OHLCV data.
    """
    print("Creating technical indicator features...")
    df = df.sort_index()

    # Simple Moving Averages (SMA)
    df['SMA_5'] = df['Close'].rolling(window=5).mean()
    df['SMA_10'] = df['Close'].rolling(window=10).mean()

    # Relative Strength Index (RSI)
    delta = df['Close'].diff(1)
    gain = delta.where(delta > 0, 0)
    loss = -delta.where(delta < 0, 0)
    avg_gain = gain.rolling(window=14).mean()
    avg_loss = loss.rolling(window=14).mean()
    rs = avg_gain / avg_loss
    df['RSI'] = 100 - (100 / (1 + rs))

    # MACD (Moving Average Convergence Divergence)
    exp1 = df['Close'].ewm(span=12, adjust=False).mean()
    exp2 = df['Close'].ewm(span=26, adjust=False).mean()
    df['MACD'] = exp1 - exp2

    # Bollinger Bands
    df['Bollinger_High'] = df['Close'].rolling(window=20).mean() + 2 * df['Close'].rolling(window=20).std()
    df['Bollinger_Low'] = df['Close'].rolling(window=20).mean() - 2 * df['Close'].rolling(window=20).std()

    # Volume-based feature
    df['Volume_SMA_5'] = df['TotalVolume'].rolling(window=5).mean()

    # Drop any rows with NaN values created by the rolling windows
    df.dropna(inplace=True)
    
    # Drop rows where volume is 0
    df = df[df['TotalVolume'] > 0]

    #df['ATR'] = np.abs(df['High'] - df['Low']).rolling(window=14).mean()

    print(f"Features created. Shape of data: {df.shape}")
    return df
```

```python
def prepare_data(data, target_column='close', look_back=60):
    """
    Scales the data and creates sequences for LSTM training.
    """
    print("Preparing data for LSTM...")
    # Select features to use
    features = ['Close', 'Open', 'High', 'Low', 'TotalVolume', 'SMA_5', 'SMA_10', 'RSI', 'MACD', 'Bollinger_High', 'Bollinger_Low', 'Volume_SMA_5']
    data = data[features].values

    # Scale the data using MinMaxScaler
    scaler = MinMaxScaler(feature_range=(0, 1))
    scaled_data = scaler.fit_transform(data)

    # Create sequences for LSTM
    X, y = [], []
    for i in range(len(scaled_data) - look_back - 1):
        X.append(scaled_data[i:(i + look_back), :])
        y.append(scaled_data[i + look_back, 0])  # Predict the 'close' price of the next day

    X, y = np.array(X), np.array(y)

    # Split into training and testing sets
    train_size = int(len(X) * 0.8)
    X_train, X_test = X[:train_size], X[train_size:]
    y_train, y_test = y[:train_size], y[train_size:]

    print(f"Training data shape: {X_train.shape}")
    print(f"Testing data shape: {X_test.shape}")

    return X_train, X_test, y_train, y_test, scaler
```

```python
def create_lstm_model(input_shape, units=50):
    """
    Builds the LSTM model architecture.
    """
    print("Creating LSTM model...")


    model = tf.keras.Sequential([
        tf.keras.layers.LSTM(units, return_sequences=True, input_shape=input_shape),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.LSTM(units, return_sequences=False),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(units=1) # Output layer for single value prediction
    ])

    
    model.compile(optimizer='adam', loss='mean_squared_error')

                
    return model
```

```python
def create_lstm_model(input_shape, units=50):
    """
    Builds the LSTM model architecture.
    """
    print("Creating LSTM model...")

    model = tf.keras.Sequential([
        tf.keras.layers.LSTM(64, return_sequences=True, input_shape=input_shape),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.LSTM(32),
        tf.keras.layers.Dropout(0.2),
        tf.keras.layers.Dense(16, activation='relu'),
        tf.keras.layers.Dense(units=1) # Output layer for single value prediction
    ])


    # Compile model
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss='mse',
        metrics=['mae']
    )            
    return model
```

```python
# --- MAIN EXECUTION ---

# Fetch and prepare the data
df = fetch_data(SYMBOL)
if df is not None and not df.empty:
    df = create_features(df)
    X_train, X_test, y_train, y_test, scaler = prepare_data(df, look_back=LOOK_BACK_DAYS)

    # Build and train the model
    model = create_lstm_model(input_shape=(X_train.shape[1], X_train.shape[2]), units=LSTM_UNITS)
    history = model.fit(X_train, y_train, epochs=EPOCHS, batch_size=BATCH_SIZE, validation_split=0.1, verbose=1)

    # --- Plotting the Model Performance (Loss) ---
    plt.figure(figsize=(10, 6))
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Model Training Performance')
    plt.xlabel('Epoch')
    plt.ylabel('Loss (Mean Squared Error)')
    plt.legend()
    plt.grid(True)
    plt.show()

    # --- Plotting the Prediction vs. Actual Prices ---
    print("Making predictions on the test set...")
    # Make predictions
    predictions = model.predict(X_test)

    # Inverse transform the predictions and actual values
    # Need to create a dummy array with the same number of features to inverse transform
    dummy_array = np.zeros((len(predictions), scaler.n_features_in_))
    dummy_array[:, 0] = predictions.flatten()
    predictions = scaler.inverse_transform(dummy_array)[:, 0]

    dummy_array = np.zeros((len(y_test), scaler.n_features_in_))
    dummy_array[:, 0] = y_test.flatten()
    actual_prices = scaler.inverse_transform(dummy_array)[:, 0]

    # Calculate RMSE
    rmse = np.sqrt(mean_squared_error(actual_prices, predictions))
    print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")

    # Plot the results
    plt.figure(figsize=(14, 7))
    plt.plot(df.index[len(df) - len(actual_prices):], actual_prices, color='blue', label='Actual SPY Price')
    plt.plot(df.index[len(df) - len(predictions):], predictions, color='red', linestyle='--', label='Predicted SPY Price')
    plt.title(f'SPY Price Prediction with LSTM')
    plt.xlabel('Date')
    plt.ylabel('SPY Price (USD)')
    plt.legend()
    plt.grid(True)
    plt.show()

else:
    print("Failed to get data. Exiting.")
```
